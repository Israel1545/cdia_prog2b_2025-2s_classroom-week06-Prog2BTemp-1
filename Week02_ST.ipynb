## Ejercicio 1 – Atributo privado + método seguro

```python

# Implementa la clase Sensor con:
# - atributo privado __valor (float)
# - método leer() que devuelve el valor
# - método actualizar(nuevo) que solo permita valores entre 0 y 100 inclusive

class Sensor:
    # TODO: tu código aquí
    def __init__(self, valor_inicial: float = 0.0):
        self.__valor = float(valor_inicial)

    def leer(self) -> float:
        return self.__valor

    def actualizar(self, nuevo: float) -> None:
        if 0 <= nuevo <= 100:
            self.__valor = float(nuevo)
        else:
            raise ValueError("Valor fuera de rango")

# TESTS (no modificar)
s = Sensor(10)
assert s.leer() == 10.0
s.actualizar(100)
assert s.leer() == 100.0
try:
    s.actualizar(101)
    raise AssertionError("Debió lanzar ValueError")
except ValueError:
    pass
print("✓ Ejercicio 1 OK")
```

## Ejercicio 2 – `@property` con validación

```python

# Crea la clase Producto con:
# - atributo protegido _precio
# - propiedad precio (getter/setter) que:
#   * asegure precio > 0
#   * redondee a 2 decimales al asignar

class Producto:
    # TODO: tu código aquí
    def __init__(self, precio: float):
        self._precio = float(precio)

    @property
    def precio(self) -> float:
        return self._precio

    @precio.setter
    def precio(self, valor: float) -> None:
        if valor <= 0:
            raise ValueError("El precio debe ser positivo")
        self._precio = round(float(valor), 2)

# TESTS (no modificar)
p = Producto(1.999)
p.precio = 3.4567
assert p.precio == 3.46
try:
    p.precio = 0
    raise AssertionError("Debió lanzar ValueError")
except ValueError:
    pass
print("✓ Ejercicio 2 OK")
```

## Ejercicio 3 – Propiedad de solo lectura derivada

```python

# Crea la clase Rectangulo con:
# - atributos protegidos _ancho, _alto
# - propiedad area (solo lectura) = ancho * alto
# - métodos set_ancho y set_alto con validación > 0

class Rectangulo:
    # TODO: tu código aquí
    def __init__(self, ancho: float, alto: float):
        self._ancho = float(ancho)
        self._alto = float(alto)

    @property
    def area(self) -> float:
        return self._ancho * self._alto

    def set_ancho(self, v: float) -> None:
        if v <= 0:
            raise ValueError("ancho debe ser > 0")
        self._ancho = float(v)

    def set_alto(self, v: float) -> None:
        if v <= 0:
            raise ValueError("alto debe ser > 0")
        self._alto = float(v)

# TESTS (no modificar)
r = Rectangulo(3, 4)
assert r.area == 12
r.set_ancho(5); r.set_alto(2)
assert r.area == 10
try:
    r.set_alto(0)
    raise AssertionError("Debió lanzar ValueError")
except ValueError:
    pass
print("✓ Ejercicio 3 OK")
```

## Ejercicio 4 – Caja fuerte: solo lectura del saldo

```python

# Implementa la clase CajaFuerte con:
# - __saldo privado
# - propiedad saldo (solo lectura)
# - método depositar(monto>0)
# - NO debe existir setter directo del saldo

class CajaFuerte:
    # TODO: tu código aquí
    def __init__(self, saldo_inicial: float = 0):
        if saldo_inicial < 0:
            raise ValueError("Saldo inicial inválido")
        self.__saldo = float(saldo_inicial)

    @property
    def saldo(self) -> float:
        return self.__saldo

    def depositar(self, monto: float) -> None:
        if monto <= 0:
            raise ValueError("monto debe ser > 0")
        self.__saldo += float(monto)

# TESTS (no modificar)
c = CajaFuerte(10)
c.depositar(2.5)
assert round(c.saldo,2) == 12.5
try:
    c.saldo = 999  # debe fallar
    raise AssertionError("No se debe poder asignar saldo directamente")
except AttributeError:
    pass
print("✓ Ejercicio 4 OK")
```

## Ejercicio 5 – Desafío: `@property` con cache


Implementa una clase `Texto` que reciba una cadena y ofrezca:
- propiedad `palabras` (lista de palabras, *lazy* con cache),
- propiedad `n_palabras` (usa `palabras`),
- si el texto cambia con `set_texto(...)`, invalida el cache.


```python

class Texto:
    # TODO: tu código aquí
    def __init__(self, contenido: str):
        self._contenido = contenido
        self._cache_palabras = None

    @property
    def palabras(self):
        if self._cache_palabras is None:
            self._cache_palabras = [w for w in self._contenido.split() if w]
        return self._cache_palabras

    @property
    def n_palabras(self):
        return len(self.palabras)

    def set_texto(self, nuevo: str):
        self._contenido = nuevo
        self._cache_palabras = None

# TESTS (no modificar)
t = Texto("hola mundo")
assert t.n_palabras == 2
t.set_texto("uno dos tres")
assert t.palabras == ["uno", "dos", "tres"]
assert t.n_palabras == 3
print("✓ Ejercicio 5 OK")
```
