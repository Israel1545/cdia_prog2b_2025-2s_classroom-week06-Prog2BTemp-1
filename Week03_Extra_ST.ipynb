**Objetivo:** Modelar una jerarquía de planes de suscripción y un flujo para calcular el costo mensual y generar un reporte por usuario.

- Clase base `Plan(nombre)` con métodos polimórficos `calcular_costo(mensualidad_base)` y `beneficios()` (lista de *str*).
- Subclases:
  - `PlanGratis`: costo = 0, beneficios limitados.
  - `PlanPro(descuento_docente=0.0)`: costo = `mensualidad_base * (1 - descuento_docente)`.
  - `PlanEnterprise(fee_soporte)`: costo = `mensualidad_base + fee_soporte`.

- **Fábrica simple:** `crear_plan(descriptor: dict)` que devuelva la instancia correcta a partir de:
  ```python
  {"tipo": "pro", "descuento_docente": 0.15}
  {"tipo": "enterprise", "fee_soporte": 25}
  {"tipo": "gratis"}
  ```
- **Mixin de logging:** `LoggerMixin` con `log(msg)`; crear `PlanConLog` que combine el mixin con una clase concreta y registre cada cálculo.
- **Uso de `super()`:** reutiliza lógica común desde la clase base.
- **Reporte:** `reporte(usuarios)` que reciba una lista de dicts como:
  ```python
  {"usuario": "ana", "plan": {"tipo": "pro", "descuento_docente": 0.1}}
  ```
