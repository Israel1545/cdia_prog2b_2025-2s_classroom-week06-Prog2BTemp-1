
## 🧠 **Ejercicio Extra – Análisis Numérico sin Visualización**

### Enunciado:
Supón que eres parte de un equipo que analiza datos experimentales de una serie de mediciones físicas.  
Cada experimento produce 1000 observaciones que siguen una distribución aproximadamente normal con media y desviación estándar diferentes en cada experimento.

1. **Genera los datos:**
   - Simula tres experimentos con `np.random.normal` usando distintas medias (`μ`) y desviaciones estándar (`σ`).
   - Guarda los resultados en una matriz `data` de tamaño `(3, 1000)`.

2. **Analiza los resultados:**
   - Calcula para cada experimento:
     - Media (`mean`)
     - Desviación estándar (`std`)
     - Valor máximo (`max`)
     - Valor mínimo (`min`)
   - Almacena los resultados en un arreglo 2D o un diccionario estructurado.

3. **Evalúa estabilidad:**
   - Calcula el **coeficiente de variación** (`std / mean`) para cada experimento.
   - Determina cuál experimento presenta la **mayor variabilidad relativa**.

4. **(Opcional avanzado)**:
   - Normaliza los tres conjuntos de datos (restar la media y dividir por la desviación estándar).
   - Verifica que cada conjunto normalizado tenga media ≈ 0 y desviación estándar ≈ 1.
