# ⭐ Ejercicio Extra (Integrador) — Gestor de Usuarios del Sistema

## Objetivo
Integrar **métodos mágicos**, **sobrecarga**, **encapsulamiento** y **Singleton**.

## Parte 1 — Clase `Usuario`
- Atributos privados: `__nombre`, `__nivel`, `__activo`  
- `__init__`, `__str__`, `__eq__` (comparación por nombre), `__add__` (aumentar nivel)

**Prueba sugerida:**
- Crear tres usuarios (dos con el mismo nombre), compararlos (`==`), aumentar nivel con `+` y mostrarlos con `print()`.

## Parte 2 — Clase `SistemaUsuarios` (Singleton)
- Única instancia mediante `__new__`
- Lista privada `__usuarios`
- `agregar_usuario(usuario)` (evitar duplicados)  
- `__len__`, `__getitem__`, `__contains__`

**Prueba sugerida:**
- Verificar que dos instancias son la misma (`is`), agregar tres usuarios (uno duplicado), usar `len()` y `in`.
